```python
class KalkulatorSederhana:
    def __init__(self, bilangan1, bilangan2):
        self.bilangan1 = bilangan1
        self.bilangan2 = bilangan2
    
    def setBilangan1(self, bilangan1):
        bilangan1.append(self.bilangan1)

    def setBilangan2(self, bilangan2):
        bilangan2.append(self.bilangan2)

    def tambah(self):
        return self.bilangan1 + self.bilangan2
    
    def pangkat(self):
        for _ in range(1, self.bilangan2):
            self.bilangan1 *= self.bilangan1
        return self.bilangan1


print("Kalkulator Sederhana")
kalkulator = KalkulatorSederhana(4, 2)
print(f"1. Penjumlahan: {kalkulator.tambah()}")
print(f"2. Pangkat: {kalkulator.pangkat()}")

```

```python
class Node:
    def __init__(self, saya):
        self.saya = saya
        self.next = None
    

class Linkedlist:
    def __init__(self, saya):
        node_baru = Node(saya)
        self.head = node_baru
        self.tail = node_baru
        self.length = 1

    def append(self, saya):
        node_baru = Node(saya)
        if self.length == 0:
            self.head = node_baru
            self.tail = node_baru
        else:
            self.tail.next = node_baru
            self.tail = node_baru
        self.length += 1
        return True
    
    def insert(self, index, saya):
        if index < 0 or index > self.length:
            return False
        if index == 0:
            return self.prepend(saya)
        if index == self.length:
            return self.append(saya)
        node_baru = Node(saya)
        temp = self.get(index - 1)
        node_baru.next = temp.next
        temp.next = node_baru
        self.length += 1
        return True

    def prepend(self, saya):
        node_baru = Node(saya)
        if self.length == 0:
            self.head = node_baru
            self.tail = node_baru
        else:
            node_baru.next = self.head
            self.head = node_baru
        self.length += 1
        return True
    
    def printList(self):
        temp = self.head
        while temp is not None:
            print(temp.saya)
            temp = temp.next

linkedlist = Linkedlist("saya")
linkedlist.append("adalah")
linkedlist.insert(2, "Awan")
linkedlist.prepend("nama")

print("Linked List:")
linkedlist.printList()

        
```
