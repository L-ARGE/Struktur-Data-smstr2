```python
import random

def swap(my_list, index1, index2):
    temp = my_list[index1]
    my_list[index1] = my_list[index2]
    my_list[index2] = temp

def pivot_random(my_list, pivot_index, end_index):
    random_index = random.randint(pivot_index, end_index)
    swap(my_list, pivot_index, random_index)

    swap_index = pivot_index
    for i in range(pivot_index+1, end_index+1):
        if my_list[i] < my_list[pivot_index]:
            swap_index += 1
            swap(my_list, swap_index, i)
    swap(my_list, pivot_index, swap_index)
    return swap_index

def random_quick_sort_helper(my_list, left, right):
    if left < right:
        pivot_index = pivot_random(my_list, left, right)
        random_quick_sort_helper(my_list, left, pivot_index-1)
        random_quick_sort_helper(my_list, pivot_index+1, right)
    return my_list

def quick_sort_random(my_list):
    random_quick_sort_helper(my_list, 0, len(my_list)-1)

my_list = [73, 94, 20, 94, 21, 43, 88, 59, 30, 48]

quick_sort_random(my_list)
print("Hasil Quick Sort Random: ")
print(my_list)



```
