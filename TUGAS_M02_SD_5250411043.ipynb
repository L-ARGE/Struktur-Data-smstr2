```python
def login():
    while True:
        masuk_npm = input("Masukkan 3 digit terakhir NPM Anda: ")
        
        if not masuk_npm.isdigit():
            print("Inputan harus berupa angka(123)")
            continue
        if len(masuk_npm) != 3:
            print("Inputan harus terdiri dari 3 digit angka")
            continue

        break
    return masuk_npm

def ubah_parameter(masuk_npm):
    list_awal = []
    for nilai in masuk_npm:
        list_awal.append(int(nilai))

    return list_awal

def perulangan_ganjil_genap(list_awal):
    list_hasil = []
    for angka in list_awal:
        if angka == 0:
            ulang = 1
        elif angka % 2 == 0:
            ulang = angka
        else:
            ulang = angka + 1

    
        for i in range(ulang):
            list_hasil.append(angka)
        
    return list_hasil

def perhitungan_total(list_hasil):
    total = 0

    for a in list_hasil:
        total += a
    return total

def perhitungan_rata(list_hasil, total):
    rata_rata = total / len(list_hasil)
    return rata_rata

def perhitungan_elemen(list_hasil, rata_rata):

    jumlah_atas = 0
    for a in list_hasil:
        if a > rata_rata:
            jumlah_atas += 1
    return jumlah_atas

def unik(list_hasil):
    list_unik = []

    for a in list_hasil:
        sudah_ada = False

        for unik in list_unik:
            if a == unik:
                sudah_ada = True
                break

        if not sudah_ada:
            list_unik.append(a)
          
    return list_unik

npm = login()
list_awal = ubah_parameter(npm)
list_hasil = perulangan_ganjil_genap(list_awal)
total = perhitungan_total(list_hasil)
rata = perhitungan_rata(list_hasil, total)
elemen = perhitungan_elemen(list_hasil, rata)
list_unik = unik(list_hasil)

print("HASIL OPERASI SEDERHANA")
print(f"List Awal               : {list_awal}")
print(f"List Hasil              : {list_hasil}")
print(f"Total Perhitungan       : {total}")
print(f"Rata - Rata Nilai       : {rata}")
print(f"Elemen > Rata - Rata    : {elemen}")
print(f"List Unik               : {list_unik}")

```
