```python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None
        self.prev = None
    
class DoublyLinkedList:
    def __init__(self, value=None):
        if value is not None:
            new_node = Node(value)
            self.head = new_node
            self.tail = new_node
            self.length = 1
        else:
            self.head = None
            self.tail = None
            self.length = 0

    def nambah_jalan(self, value):

        new_node = Node(value)
        if self.head is None:
            self.head = new_node
            self.tail = new_node
        else:
            self.tail.next = new_node
            new_node.prev = self.tail
            self.tail = new_node
        self.length += 1
        return True

    def perjalanan_berangkat(self):
        temp = self.head
        while temp is not None:
            print(temp.value, end=' => ')
            temp = temp.next
        print('Tugu Jogja')

    def perjalanan_pulang(self):
        temp = self.head

        while temp.next:
            temp = temp.next
        
        while temp:
            print(temp.value, end=' => ')
            temp = temp.prev
        print('Kost')

rute = DoublyLinkedList()
rute.nambah_jalan("Jln. Mlati")
rute.nambah_jalan("Jl. Kb. Agung Raya")
rute.nambah_jalan("Jl. Magelang")
rute.nambah_jalan("Jl. Margo Utomo")

print("Rute berangkat:")
rute.perjalanan_berangkat()
print("\nRute pulang:")
rute.perjalanan_pulang()
```
