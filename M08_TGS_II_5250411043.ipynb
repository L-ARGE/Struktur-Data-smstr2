```python
class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None
    
class BinarySearchTree:
    def __init__(self):
        self.root = None

    def insert(self, value):
        new_node = Node(value)
        if self.root is None:
            self.root = new_node
            return True
        temp = self.root
        while (True):
            if new_node.value == temp.value:
                return False
            if new_node.value < temp.value:
                if temp.left is None:
                    temp.left = new_node
                    return True
                temp = temp.left
            else:
                if temp.right is None:
                    temp.right = new_node
                    return True
                temp = temp.right
    
    def contains(self, value):
        temp = self.root
        while (temp is not None):
            if value < temp.value:
                temp = temp.left
            elif value > temp.value:
                temp = temp.right
            else:
                return True
        return False    
        
    def min_value_node(self, current_node):
        while current_node.left is not None:
            current_node = current_node.left
        return current_node
            
    def maxValue(self, current_node):
        while current_node.right is not None:
            current_node = current_node.right
        return current_node
    
    def averageValue(self, node):
        if node is None:
            return (0, 0)
            
        total_kiri, nilai_kiri = self.averageValue(node.left)
        total_kanan, nilai_kanan = self.averageValue(node.right)
            
        total = node.value + total_kiri + total_kanan
        rata_rata = 1 + nilai_kiri + nilai_kanan

        return(total, rata_rata)
        
    def print_averageValue(self):
        total, rata_rata = self.averageValue(self.root)

        if rata_rata == 0:
            return 0
        
        return total/rata_rata
    
    def informasi_tree(self, node):
        if node is None :
            return 0, 0, 0
        
        total_kiri, total_varian_kiri, nilai_kiri = self.informasi_tree(node.left)
        total_kanan, total_varian_kanan, nilai_kanan = self.informasi_tree(node.right)

        total_jumlah = node.value + total_kiri + total_kanan
        total_varian = (node.value ** 2) + total_varian_kiri + total_varian_kanan
        total_nilai = 1 + nilai_kiri + nilai_kanan

        return total_jumlah, total_varian, total_nilai
    
    def nilai_deviasi(self):
        total_jumlah, total_varian, total_nilai = self.informasi_tree(self.root)

        if total_nilai == 0:
            return  0
        
        mean = total_jumlah / total_nilai
        variasi = (total_varian / total_nilai) - (mean ** 2)

        return (max(0, variasi)) ** 0.5
    
    def hapus(self, value):
        self.root = self._hapus_node(self.root, value)

    def _hapus_node(self, node, value):
        if node is None:
            return None
        
        if value < node.value:
            node.left = self._hapus_node(node.left, value)
        elif value > node.value:
            node.right = self._hapus_node(node.right, value)
        else:
            if node.left is None:
                return node.right
            elif node.right is None:
                return node.left
            
            temp = self.min_value_node(node.right)
            node.value = temp.value
            node.right = self._hapus_node(node.right, temp.value)
        return node
    
    def ambil_data(self, node, data_list):
        if node is None:
            return
        self.ambil_data(node.left, data_list)
        data_list.append(node.value)
        self.ambil_data(node.right, data_list)
        return data_list

    def penyeimbang(self, data_list, awal, akhir):
        if awal > akhir:
            return None
        
        middle = (awal + akhir) // 2
        node = Node(data_list[middle])
        node.left = self.penyeimbang(data_list, awal, middle - 1)
        node.right = self.penyeimbang(data_list, middle + 1, akhir)
        return node
    
    def reconstruct_tree(self):
        data_list = []
        self.ambil_data(self.root, data_list)
        self.root = self.penyeimbang(data_list, 0, len(data_list) - 1)

my_tree = BinarySearchTree()
my_tree.insert(47)
my_tree.insert(21)
my_tree.insert(76)
my_tree.insert(18)
my_tree.insert(27)
my_tree.insert(52)
my_tree.insert(82)

print('Posisi Root/Akar:')
print("Root:", my_tree.root.value)
print('='*25)

print('Posisi Anak:')
print("Root -> Left:", my_tree.root.left.value)
print("Root -> Right:", my_tree.root.right.value)
print('='*25)

print('BST Contains 27:')
print(my_tree.contains(27))
print('='*25)

print('BST Contains 17:')
print(my_tree.contains(17))
print('='*25)

print('Minimum value in Tree:')
print(my_tree.min_value_node(my_tree.root).value)
print('='*25)

print('Maximum value in Tree:')
print(my_tree.maxValue(my_tree.root).value)
print('='*25)


print('Average:')
print(my_tree.print_averageValue())
print('='*25)

print('Standar Deviasi:')
print(my_tree.nilai_deviasi())
print('='*25)

print('Setelah menghapus 47 & sebelum reconstruct:')
my_tree.hapus(47)
print("Root:", my_tree.root.value)
print('='*25)

print('Setelah reconstruct:')
my_tree.reconstruct_tree()
print("Root:", my_tree.root.value)
```
