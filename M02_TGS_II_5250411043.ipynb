```python
def hitung_bunga_final(npm):
    digit9 = int(npm[8])
    digit10 = int(npm[9])
    digit_terakhir = int(npm[-1])

    bunga_final = digit9 + digit10 + (digit_terakhir % 3)

    if bunga_final < 5:
        return 5
    elif bunga_final > 20:
        return 20
    
    return bunga_final

def login_npm():

    while True:
        npm = input("Masukkan 10 digit NPM Anda: ")

        if not npm.isdigit():
            print("Inputan harus berupa angka")
        elif len(npm) != 10:
            print("Inputan harus harus berjumlah 10 digit")
        else:
            print("Inputan yang anda masukkan adalah", npm)
            return npm
        

def diskon_uang_muka(npm, uang_muka):
    digit_akhir = int(npm[9])

    if digit_akhir % 2 == 0:
        diskon = 0.1
    else:
        diskon = 0.05

    jumlah_diskon = uang_muka * diskon
    uang_muka_baru = uang_muka - jumlah_diskon
    
    return uang_muka_baru, jumlah_diskon, diskon

def hitung_simulasi_kredit(harga_barang, uang_muka, bunga_final, tahun_kredit):
    jumlah_bulan = tahun_kredit * 12
    jumlah_pinjaman = harga_barang - uang_muka

    bunga_bulanan = (bunga_final / 100) / 12
    
    cicilan_bulanan = jumlah_pinjaman * (
        bunga_bulanan / (1 - (1 + bunga_bulanan) ** (-jumlah_bulan))
    )

    total_pembayaran = cicilan_bulanan * jumlah_bulan

    return  cicilan_bulanan, total_pembayaran

def simulasi_tenor(harga_barang, uang_muka, bunga_final):
    hasil = []

    for tenor in range(1, 5):
        cicilan, total_pembayaran = hitung_simulasi_kredit(harga_barang,
                                                uang_muka,
                                                bunga_final,
                                                tenor)
        
        jumlah_pinjaman = harga_barang - uang_muka
        total_bunga = total_pembayaran - jumlah_pinjaman

        data = {
            "tenor" : tenor,
            "cicilan" : cicilan,
            "total bayar" : total_pembayaran,
            "total bunga" : total_bunga
        }

        hasil.append(data)
    return hasil

def rupiah(angka):
    return f"Rp {(angka):,.0f}"

def tampilkan_tabel(data_simulasi):
    print("")
    print("SIMULASI KREDIT ")
    print("=" * 16)

    print("")
    print("Tenor | Cicilan          | Total Bayar   | Total Bunga")
    print("=" * 55)
    for data in data_simulasi:
        print(
            f"{data['tenor']}     | "
            f"{rupiah(data['cicilan'])}     | "
            f"{rupiah(data['total bayar'])} | "
            f"{rupiah(data['total bunga'])}"
        )

npm = login_npm()
bunga = hitung_bunga_final(npm)

uang_muka = 5000000
harga_barang = 32999000
tahun_kredit = 4
total_peminjaman = harga_barang - uang_muka

simulasi_awal = simulasi_tenor(harga_barang, uang_muka, bunga)

print("")
print("INFORMASI KREDIT")
print("=" * 17)
print("Bunga Final        :", bunga, "%")
print("DP Awal            :", rupiah(uang_muka))
print("Total Harga Barang :", rupiah(harga_barang))
print("Total Tahun Kredit :", tahun_kredit, "Tahun")
tampilkan_tabel(simulasi_awal)

uang_muka_baru, jumlah_diskon, persen_diskon = diskon_uang_muka(npm, uang_muka)

print("")
print("SETELAH DISKON UANG MUKA")
print("=" * 18)
print("Diskon DP          :", int(persen_diskon * 100), "%")
print("Jumlah Diskon      :", rupiah(jumlah_diskon))
print("DP Setelah Diskon  :", rupiah(uang_muka_baru))

simulasi_diskon = simulasi_tenor(harga_barang, uang_muka_baru, bunga)
tampilkan_tabel(simulasi_diskon)
```
