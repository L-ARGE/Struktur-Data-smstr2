```python
#LATIHAN SOAL NOMER 1

Nama_lengkap = input("Masukkan nama lengkap anda: ")
npm = int(input("Masukkan NPM milik anda: "))
Tanggal_lahir = (input("Masukkan tanggal(00)-bulan(00)-tahun(0000) kelahiran anda dengan format angka: "))

tanggal = int(Tanggal_lahir[0:2])
bulan = int(Tanggal_lahir[3:5])
tahun = int(Tanggal_lahir[6:10])

tanggal_hari_ini = 10
bulan_hari_ini = 2
tahun_hari_ini = 2026

usia = tahun_hari_ini - tahun
if (bulan, tanggal) > (bulan_hari_ini,tanggal_hari_ini):
    usia -= 1

print(f"Hai, {Nama_lengkap} NPM {npm}! Pada hari ini Selasa 10 Februari 2026 Anda berumur {usia} tahun. Selamat belajar Struktur Data!")
```
