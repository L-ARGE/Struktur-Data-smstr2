```python
# 10.5 Tugas 

class Graph:
    def __init__(self):
        self.adj_list = {}

    def print_graph(self):
        for vertex in self.adj_list:
            print(vertex, ':', self.adj_list[vertex])

    def add_vertex(self, vertex):
        if vertex not in self.adj_list.keys():
            self.adj_list[vertex] = []
            return True
        return False
    
    def add_edge(self, v1, v2):
        if v1 in self.adj_list.keys() and v2 in self.adj_list.keys():
            self.adj_list[v1].append(v2)
            self.adj_list[v2].append(v1)
            return True 
        return False
    
    def remove_edge(self, v1, v2):
        if v1 in self.adj_list.keys() and v2 in self.adj_list.keys():
            try:
                self.adj_list[v1].remove(v2)
                self.adj_list[v2].remove(v1)
            except ValueError:
                pass
            return True
        return False

    def remove_vertex(self, vertex):
        if vertex in self.adj_list.keys():
            for other_vertex in self.adj_list[vertex]:
                self.adj_list[other_vertex].remove(vertex)
            del self.adj_list[vertex]
            return True
        return False
    
    def isInGraph(self, vertex):
        if vertex in self.adj_list.keys():
            return True
        return False


    def printAllConnected(self, vertex):
        if vertex in self.adj_list.keys():
            print(f"Vertex yang terhubung dengan {vertex}: {self.adj_list[vertex]}")
        else:
            print(f"Vertex {vertex} adalah vertex yang tidak ditemukan dalam graph.")

my_graph = Graph()
my_graph.add_vertex(82)
my_graph.add_vertex(27)
my_graph.add_vertex(52)
my_graph.add_vertex(18)
my_graph.add_vertex(21)
my_graph.add_vertex(76)
my_graph.add_vertex(44)

my_graph.add_edge(82, 27)
my_graph.add_edge(27, 52)
my_graph.add_edge(52, 18)
my_graph.add_edge(18, 21)
my_graph.add_edge(18, 44)
my_graph.add_edge(21, 76)
my_graph.add_edge(44, 76)
my_graph.add_edge(44, 21)

print('Graph before remove_vertex():')
my_graph.print_graph()

my_graph.remove_vertex(44)

print('\nGraph after remove_vertex():')
my_graph.print_graph()

print('\nTest fungsi IsInGraph():')
print(my_graph.isInGraph(82))
print(my_graph.isInGraph(27))
print(my_graph.isInGraph(52))
print(my_graph.isInGraph(18))
print(my_graph.isInGraph(21))
print(my_graph.isInGraph(76))
print(my_graph.isInGraph(44))

print('\nTest fungsi printAllConnected():')
my_graph.printAllConnected(82)
my_graph.printAllConnected(27)
my_graph.printAllConnected(52)
my_graph.printAllConnected(18)
my_graph.printAllConnected(21)
my_graph.printAllConnected(76)
my_graph.printAllConnected(44)
```
