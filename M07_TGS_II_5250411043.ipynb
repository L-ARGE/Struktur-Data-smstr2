```python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None

class Queue:
    def __init__(self, value):
        new_node = Node(value)
        self.first = new_node
        self.last = new_node
        self.length = 1
            
    def print(self):
        temp = self.first
        while temp is not None:
            print(temp.value)
            temp = temp.next

    def enqueue(self, value):
        new_node = Node(value)
        if self.first is None:
            self.first = new_node
            self.last = new_node
        else:
            self.last.next = new_node
            self.last = new_node
        self.length += 1

    def dequeue(self):
        if self.length == 0:
            return None
        temp = self.first
        if self.length == 1:
            self.first = None
            self.last = None
        else:
            self.first = self.first.next
            temp.next = None
        self.length -= 1
        return temp

#Fungsi desperation
    def desperation(self, x):
        if x < 1 or x > self.length:
            return None
        
        posisi = self.length - x
        if posisi == 0:
            return self.dequeue()
        
        depan = self.first
        for _ in range(posisi - 1):
            depan = depan.next

        temp = depan.next
        depan.next = temp.next
        
        if temp == self.last:
            self.last = depan

        temp.next = None
        self.length -= 1
        return temp.value

# Fungsi swap_middle
    def swap_middle(self):
        if self.length == 0:
            return None
        if self.length % 2 == 1:
            tengah = self.length // 2
            
            if tengah == 0:
                return self.dequeue()
            
            prev = self.first
            for _ in range(tengah - 1):
                prev = prev.next

            temp = prev.next
            prev.next = temp.next

            if temp == self.last:
                self.last = prev
            
            temp.next = None
            self.length -= 1
            return temp.value
        else:
            tengah1 = self.length // 2 - 1
            tengah2 = self.length // 2
            
            prev1 = self.first
            for _ in range(tengah1 - 1):
                prev1 = prev1.next

            prev2 = self.first
            for _ in range(tengah2 - 1):
                prev2 = prev2.next

            temp1 = prev1.next
            temp2 = prev2.next

            prev1.next, prev2.next = temp2, temp1

            if temp1 == self.last:
                self.last = prev2
            if temp2 == self.last:
                self.last = prev1
            
            temp1.next, temp2.next = temp2.next, temp1.next
            return (temp1.value, temp2.value)



my_queue = Queue(10)
my_queue.enqueue(20)
my_queue.enqueue(30)
my_queue.enqueue(40)
my_queue.enqueue(50)
my_queue.enqueue(60)
print("Queue awal:")
my_queue.print()

print("\nSetelah di desperation (2):")
my_queue.desperation(1)
my_queue.print()

print("\nSetelah swap middle:")
my_queue.swap_middle()
my_queue.print()

```
