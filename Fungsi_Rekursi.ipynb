```python
# Memanggil fungsi Call Stack
def funcThree():
    print("Three")
def funcTwo():
    funcThree()
    print("Two")
def funcOne():
    funcTwo()
    print("One")

funcOne()
```

```python
# 11.3.2 Kasus 1 - Faktorial
def factorial(n):
    if n == 1:
        return 1
    return n * factorial(n - 1)

print(factorial(5))
```
