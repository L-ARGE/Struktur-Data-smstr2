```python
#OPSI 1
#1 Menghapus semua elemen yang nilainya dibawah dari 20
class Node:
    def __init__(self, angka):
        self.angka = angka
        self.next = None

class Linkedlist:
    def __init__(self):
        self.head = None
    
    def tambah(self, angka):
        node_baru = Node(angka)

        if not self.head:
            self.head = node_baru
            return
        
        temp = self.head
        while temp.next:
            temp = temp.next
        temp.next = node_baru

    def hapus_bawah_20(self):
        while self.head and self.head.angka < 20:
            self.head = self.head.next
        
        temp = self.head
        while temp and temp.next:
            if temp.next.angka < 20:
                temp.next = temp.next.next
            else:
                temp = temp.next
        
    def tampil(self):
        temp = self.head
        while temp:
            print(temp.angka, end=' ')
            temp = temp.next
        print()

list1 = Linkedlist()
data = [23,43,12,71,87]

for i in data:
    list1.tambah(i)

print('list awal sebelum ada perubahan:')
list1.tampil()

print('\nsetelah dihapus untuk nilai dibawah 20:')
list1.hapus_bawah_20()
list1.tampil()
```

```python
#OPSI 1
#2 Merverse /membalik angka linked list
class Node:
    def __init__(self, angka):
        self.angka = angka
        self.next = None

class Linkedlist:
    def __init__(self):
        self.head = None
    
    def tambah(self, angka):
        node_baru = Node(angka)

        if not self.head:
            self.head = node_baru
            return
        
        temp = self.head
        while temp.next:
            temp = temp.next
        temp.next = node_baru

    def hapus_bawah_20(self):
        while self.head and self.head.angka < 20:
            self.head = self.head.next
        
        temp = self.head
        while temp and temp.next:
            if temp.next.angka < 20:
                temp.next = temp.next.next
            else:
                temp = temp.next
        
    def reversed(self):
        prev = None
        sekarang = self.head
        while sekarang:
            next_node = sekarang.next
            sekarang.next = prev
            prev = sekarang
            sekarang = next_node
        self.head = prev

    def tampil(self):
        temp = self.head
        while temp:
            print(temp.angka, end=' ')
            temp = temp.next
        print()

list1 = Linkedlist()
data = [23,43,12,71,87]

for i in data:
    list1.tambah(i)
    
print('list awal sebelum ada perubahan:')
list1.tampil()

print('\nsesudah dihapus angka dibawah 20 dan sebelum direverse:')
list1.hapus_bawah_20()
list1.tampil()

print('\nsetelah dibalik/reverse:')
list1.reversed()
list1.tampil()
```

```python
#OPSI 2
#2 Partial Reverse / Reverse dalam Range 
class Node:
    def __init__(self, angka):
        self.angka = angka
        self.next = None

class Linkedlist:
    def __init__(self):
        self.head = None
    
    def tambah(self, angka):
        node_baru = Node(angka)

        if not self.head:
            self.head = node_baru
            return
        
        temp = self.head
        while temp.next:
            temp = temp.next
        temp.next = node_baru

    def partial_reversed (self, start_idx, end_idx):
        if start_idx >= end_idx:
            return
        
        awal = Node(0)
        awal.next = self.head
        prev = awal

        for _ in range(start_idx):
            prev = prev.next
        
        sekarang = prev.next
        for _ in range(end_idx - start_idx):
            next_node = sekarang.next
            sekarang.next = next_node.next
            next_node.next = prev.next
            prev.next = next_node
        
        self.head = awal.next

    def tampil(self):
        temp = self.head
        while temp:
            print(temp.angka, end=' ')
            temp = temp.next
        print()

list1 = Linkedlist()
data = [23,43,12,71,87]

for i in data:
    list1.tambah(i)
    
print('sebelum dirubah/reverse:')
list1.tampil()

print('\nsetelah dibalik/reverse:')
list1.partial_reversed(1, 3)
list1.tampil()
```

```python
#Opsi 3
#Reverse & Split (Ganjil dan Genap)
class Node:
    def __init__(self, angka):
        self.angka = angka
        self.next = None

class Linkedlist:
    def __init__(self):
        self.head = None
    
    def tambah(self, angka):
        node_baru = Node(angka)

        if not self.head:
            self.head = node_baru
            return
        
        temp = self.head
        while temp.next:
            temp = temp.next
        temp.next = node_baru

    def genap(self):
        prev = None
        sekarang = self.head
        while sekarang:
            if sekarang.angka % 2 == 0:
                if prev:
                    prev.next = sekarang.next
                    sekarang.next = self.head
                    self.head = sekarang
                    sekarang = prev.next
                else:
                    prev = sekarang
                    sekarang = sekarang.next
            else:
                prev = sekarang
                sekarang = sekarang.next

    def reverse_split(self):
        prev = None
        sekarang = self.head
        while sekarang:
            next_node = sekarang.next
            sekarang.next = prev
            prev = sekarang
            sekarang = next_node
        self.head = prev

        
    def tampil(self):
        temp = self.head
        while temp:
            print(temp.angka, end=' ')
            temp = temp.next
        print()

list1 = Linkedlist()
data = [23,43,12,71,87]

for i in data:
    list1.tambah(i)
print('tampilan list sebelum dirubah:')
list1.tampil()
    
print('\nsebelum direverse (genap di depan):')
list1.genap()
list1.tampil()

print('\nsetelah dibalik/reverse (genap dibelakang):')
list1.reverse_split()
list1.tampil()
```
