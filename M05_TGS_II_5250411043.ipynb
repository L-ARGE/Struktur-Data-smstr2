```python
class Node:
    def __init__(self, saya):
        self.saya = saya
        self.next = None

class Linkedlist:
    def __init__(self, saya):
        node_baru = Node(saya)
        self.head = node_baru
        self.tail = node_baru
        self.length = 1

    def append(self, saya):
        node_baru = Node(saya)
        if self.length == 0:
            self.head = node_baru
            self.tail = node_baru
        else:
            self.tail.next = node_baru
            self.tail = node_baru
        self.length += 1
        return True
    
    def get(self, index):
        if index < 0 or index >= self.length:
            return None
        temp = self.head
        for _ in range(index):
            temp = temp.next
        return temp
    
    def set_value(self, index, saya):
        temp = self.get(index)
        if temp:
            temp.saya = saya
            return True
        return False

    def insert(self, index, saya):
        if index < 0 or index > self.length:
            return False
        if index == 0:
            return self.prepend(saya)
        if index == self.length:
            return self.append(saya)
        
        node_baru = Node(saya)
        temp = self.get(index - 1)
        node_baru.next = temp.next
        temp.next = node_baru
        self.length += 1
        return True

    def prepend(self, saya):
        node_baru = Node(saya)
        if self.length == 0:
            self.head = node_baru
            self.tail = node_baru
        else:
            node_baru.next = self.head
            self.head = node_baru
        self.length += 1
        return True
    
    def printList(self):
        temp = self.head
        while temp is not None:
            print(temp.saya)
            temp = temp.next

print('sebelum diganti:')
linkedlist = Linkedlist("saya")
linkedlist.append("adalah")
linkedlist.insert(2, "Mikhael Kurniawan")
linkedlist.prepend("nama")
linkedlist.printList()
        
print('\nsesudah diganti:')
linkedlist.set_value(3, "Bumi Juan Santoso")
linkedlist.printList()
        
```
