```python
class ProfilMahasiswa:
    def __init__ (self, nama_lengkap, riwayat_nilai):
        self.nama_lengkap = nama_lengkap
        self.riwayat_nilai = riwayat_nilai
        
    def catatan_nilai_baru(self, nilai_baru):
        self.riwayat_nilai.append(nilai_baru)
        
    def tampilkan_transkip(self):
        print(f'Nama Mahasiswa   : {self.nama_lengkap}')
        print(f'Riwayat Nilai    : {self.riwayat_nilai}')
        
mahasiswa_asli = ProfilMahasiswa("Mikhael Kurniawan", [])

mahasiswa_asli.catatan_nilai_baru(80)

mahasiswa_asli.tampilkan_transkip()

referensi_mahasiswa = mahasiswa_asli

referensi_mahasiswa.catatan_nilai_baru(95)

mahasiswa_asli.tampilkan_transkip()

print(f'Id mahasiswa asli       : {id(mahasiswa_asli)}')
print(f'Id refernsi mahasiswa   : {id(referensi_mahasiswa)}')#
```

Dapat disimpulkan bahwa variable referensi_mahasiswa memiliki id yang sama dengan id miliki mahasiswa asli. Hal ini dapat terjadi dikarenakan variable referensi_mahasiswa menunjuk kepada objek yang sama dengan yang ada dimemori. Sehingga nilai yang ada didalam variable referensi_mahasiswa masuk kedalam list dari variable riwayat_nilai yang sebelumnya sudah diisi oleh variabel dengan nama mahasiswa_asli
