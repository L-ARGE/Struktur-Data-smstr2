```python
stok_pusat = {
    "Elektronik": 
    {
        "Laptop": 10,
        "Mouse": 15
        },
    "ATK": 
    {
        "Kertas": 50,
        "Pena": 100
        }
}

stok_cabang_A = stok_pusat
stok_cabang_A["ATK"]["Kertas"] = 30

print(130*'=')
print("Stok Pusat(Awal) : ", stok_pusat)
print("Stok cabang A(sama dengan stok pusat) : ", stok_cabang_A)
print(130*'-')

print(130*'-')
print("Id Stok Pusat : ", id(stok_pusat))
print("Id Stok Cabang A(sama): ", id(stok_cabang_A))
print(130*'=')

stok_cabang_B = stok_pusat.copy()
stok_cabang_B["Furnitur"] = {"Meja": 5}

print(130*'=')
print("Stok cabang B : ", stok_cabang_B)
print("Stok Pusat(tidak terpengaruh perubahan stok cabang B) : ", stok_pusat)
print(130*'-')

print(130*'-')
print("Id Stok Pusat : ", id(stok_pusat))
print("Id Stok Cabang B(berbeda): ", id(stok_cabang_B))

print(130*'-')
stok_cabang_B["Elektronik"]["Laptop"] = 99

print("Stok cabang B(setelah perubahan): ",stok_cabang_B)
print("Stok Pusat(Akhir) : ", stok_pusat)
print(130*'=')

stok_cabang_C = {}

for kategori in stok_pusat:
    stok_cabang_C[kategori] = stok_pusat[kategori].copy()

print("Stok cabang C(awal) :", stok_cabang_C)

stok_cabang_C["Elektronik"]["Laptop"] = 10

print("Stok cabang C(akhir) :", stok_cabang_C)
print("Stok pusat(tidak ikut berubah) :",stok_pusat)

print("Id Stok cabang C  : ", id(stok_cabang_C))
print("Id Stok Pusat(tidak ikut berubah) : ", id(stok_pusat))
print(130*'=')
```

Kesimpulan nomor 3:
Isi dict ATK untuk kertas berubah menjadi 30 dikarenakan pada variabel stok cabang A di ganti menjadi 30, hal itu juga otomatis berubah pada stok cabang pusat juga dikarenakan stok cabang A mengambil isi data dari stok pusat dan ini juga mengacu kepada id ke dua variabel yang sama dikarenakan stok cabang A hanya mengambil dari isi data stok pusat yang menjadikan id kedua variabel sama
Kesimpulan nomor 5:
Jumlah "Laptop" pada stok pusat ikut berubah dikarenakan dictionary tersebut juga ada pada stok cabang B jadi ikut berubah, sedangkan mengapa kategori "Furnitur" berhasil dipisahkan dikarenakan kategori furnitur ditambahkan pada variabel stok cabang B alhasil stok pusat tidak terpengaruh
