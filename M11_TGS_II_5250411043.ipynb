```python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None

class LinkedList:
    def __init__(self):
        self.head = None

    def append(self, value):
        node_baru = Node(value)
        if not self.head:
            self.head = node_baru
            return
        current_node = self.head
        while current_node.next:
            current_node = current_node.next
        current_node.next = node_baru

    def print_list(self):
        hasil = []
        current= self.head
        while current:
            hasil.append(str(current.value))
            current = current.next
        print(hasil)

    def asc_linked_list(self):
        if self.head is None:
            return
        
        pengurutan = False
        while not pengurutan:
            pengurutan = True
            current_node = self.head
            while current_node.next:
                if current_node.value > current_node.next.value:
                    temp = current_node.value
                    current_node.value = current_node.next.value
                    current_node.next.value = temp
                    pengurutan = False
                current_node = current_node.next

    def dsc_linked_list(self):
        if self.head is None:
            return
        
        pengurutan = False
        while not pengurutan:
            pengurutan = True
            current_node = self.head
            while current_node.next:
                if current_node.value < current_node.next.value:
                    temp = current_node.value
                    current_node.value = current_node.next.value
                    current_node.next.value = temp
                    pengurutan = False
                current_node = current_node.next


listku = LinkedList()
listku.append(145)
listku.append(178)
listku.append(167)
listku.append(609)
listku.append(254)
listku.append(921)
listku.append(549)
listku.append(492)
listku.append(723)
listku.append(316)
listku.append(835)


print("=" * 78)
print("\t \t \t Linked List sebelum diurutkan:")
print("=" * 78)
listku.print_list()

print("\n" + "=" * 78)
print("\tLinked List setelah diurutkan secara ascending (kecil > besar):")
print("=" * 78)
listku.asc_linked_list()
listku.print_list()

print("\n" + "=" * 78)
print("\tLinked List setelah diurutkan secara descending (besar > kecil):")
print("=" * 78)
listku.dsc_linked_list()
listku.print_list()
```
