```python
def login():
    while True:
        masuk_npm = input("Masukkan 3 digit terakhir NPM Anda: ")

        if not masuk_npm.isdigit():
            print("Inputan harus berupa angka(123)")
            continue
        if len(masuk_npm) != 3:
            print("Inputan harus terdiri dari 3 digit angka")
            continue

        break
    return masuk_npm

def ubah_parameter(masuk_npm):
    list_awal = []
    for nilai in masuk_npm:
        list_awal.append(int(nilai))

    return list_awal

def perulangan_ganjil_genap(list_awal):
    list_hasil = []
    for angka in list_awal:
        if angka == 0:
            ulang = 1
        elif angka % 2 == 0:
            ulang = angka
        else:
            ulang = angka + 1


        for i in range(ulang):
            list_hasil.append(angka)

    return list_hasil

def perhitungan_total(list_hasil):
    total = 0

    for a in list_hasil:
        total += a
    return total

def perhitungan_rata(list_hasil, total):
    rata_rata = total / len(list_hasil)
    return rata_rata

def perhitungan_elemen(list_hasil, rata_rata):

    jumlah_atas = 0
    for a in list_hasil:
        if a > rata_rata:
            jumlah_atas += 1
    return jumlah_atas

def unik(list_hasil):
    list_unik = []

    for a in list_hasil:
        sudah_ada = False

        for unik in list_unik:
            if a == unik:
                sudah_ada = True
                break

        if not sudah_ada:
            list_unik.append(a)

    return list_unik

npm = login()
list_awal = ubah_parameter(npm)
list_hasil = perulangan_ganjil_genap(list_awal)
total = perhitungan_total(list_hasil)
rata = perhitungan_rata(list_hasil, total)
elemen = perhitungan_elemen(list_hasil, rata)
list_unik = unik(list_hasil)

print("HASIL OPERASI SEDERHANA")
print(f"List Awal               : {list_awal}")
print(f"List Hasil              : {list_hasil}")
print(f"Total Perhitungan       : {total}")
print(f"Rata - Rata Nilai       : {rata}")
print(f"Elemen > Rata - Rata    : {elemen}")
print(f"List Unik               : {list_unik}")

```

def login()

def ubah_parameter(masuk_npm)
Fungsi ini digunakan untuk mengubah inputan awal dari sebuah string menjadi integer lalu memasukkan angka inputan kedalam list (list_awal)

def perulangan_ganjil_genap(list_awal)
Ini digunakan untuk melakukan perulangan dan perhitungan terkait nilai yang dimasukkan. Dimana ketika terdapat nilai genap, maka angka inputan akan diulang sebanyak nilai genap tersebut sedangakan untuk nilai ganjil perulangan akan diulang sesuai nilainya namun di tambah 1 kali. Lalu hasilnya akan di simpan pada list_hasil.

def perhitungan_total(list_hasil)
Fungsi ini digunakan untuk melakukan perhitungan total penjumlahan dari nilai pada hasil list_hasil yang akan disimpan pada variable total.

def perhitungan_rata(list_hasil, total):
Fungsi ini digunakan untuk menghitung hasil rata rata dari hasil penjumlahan yang tersimpan di variable total serta akan disimpan di variable rata-rata. Perhitungan rata-rata ini dilakukan dengan menjumlah semua nilai perulangan lalu membaginya dengan jumlah nilai yang ada dengan atribut len().

def perhitungan_elemen(list_hasil, rata_rata)
Ini digunakan melakukan perhitungan dimana apakah nilai yang ada pada list_hasil ada yang lebih besar (>) dari hasil rata rata yang telah dihitung sebelumnya.

def unik(list_hasil)
Ini digunakan untuk melakukan cek nilai unik. Dimana nilai unik sendiri adalah nilai yang muncul sekali dari suatu list yang ada.

Kesimpulannya source code yang ada ini berjalan dengan memasukkan 3 digit angka terakhir dari NPM milik mahasiswa, dimana jika terdapat kesalahan pada inputannya maka akan diulangi lagi.

Lalu nilai yang sudah dimasukkan akan diubah tipenya menjadi int untuk dimasukkan kedalam list. Lalu list itu akan dilakukan perulangan dimana jika terdapat angka genap maka akan dilakukan perulangan dengan nilai yang sama, namun jika itu ganjil maka nilai ganjil itu akan ditambah 1.

Selanjutnya akan dilakukan penjumlahan dari list perulangan sebelumnya, juga dilakukan perhitungan rata-rata men untuk menentukan nilai elemen mana yang lebih tinggi dari rata-rata.

Lalu sebelum keakhir, akan dilakukan juga perhitungan nilai unik dari list yang telah dibuat.

Setelah semuanya dilakukan maka pada akhirnya akan dilakukan pemanggilan fungsi dari apa yang telah dibuat dan akan dilakukan eksekusi di akhirnya.
